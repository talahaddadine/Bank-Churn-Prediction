# 5) Data Splitting

```python
from sklearn.model_selection import train_test_split
X = bank.drop('Attrition_Flag', axis=1)
y = bank['Attrition_Flag']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

# 5) Building three models

# 5.1 K-NN Algorithm

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
```

```python
knn = KNeighborsClassifier(n_neighbors = 15 ,weights='distance')
knn.fit(X_train,y_train)
knn_clf = KNeighborsClassifier() 
knn_clf.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
sns.heatmap(cm,annot=True,fmt='d',linecolor = 'white',linewidths=5)
```

```python
acc_k_15 = accuracy_score(y_test,y_pred)
print('the accuracy of the model is:',acc_k_15*100,'%')
```

```python
print(classification_report(y_test,y_pred))
```

```python
knn5 = KNeighborsClassifier(n_neighbors = 5 ,weights='distance')
knn.fit(X_train,y_train)
knn_clf = KNeighborsClassifier() 
knn_clf.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
sns.heatmap(cm,annot=True,fmt='d',linecolor = 'white',linewidths=5)
```

```python
acc_k_5 = accuracy_score(y_test,y_pred)
print('the accuracy of the model is:',acc_k_5*100,'%')
```

```python
print(classification_report(y_test,y_pred))
```

```python
knn = KNeighborsClassifier(n_neighbors = 3 ,weights='distance')
knn.fit(X_train,y_train)
knn_clf = KNeighborsClassifier() 
knn_clf.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
cm = confusion_matrix(y_pred,y_test)
sns.heatmap(cm,annot=True,fmt='d',linecolor = 'white',linewidths=5)
```

```python
acc_k_3 = accuracy_score(y_test,y_pred)
print('the accuracy of the model is:',acc_k_3*100,'%')
```

```python
print(classification_report(y_test,y_pred))
```

```python
scores = [acc_k_3*100,acc_k_5*100,acc_k_15*100]
Models = pd.DataFrame({
    'Model': ["k=3","k=5","k=15"],
    'Score (%)': scores})
Models.sort_values(by='Score (%)', ascending=False)
```
